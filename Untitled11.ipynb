```python
import os
import sys
import subprocess
from pathlib import Path
from google.colab import drive

# ==============================================================================
# 1. MOUNT GOOGLE DRIVE & DEKLARASI PATH SPESIFIK
# ==============================================================================
drive.mount('/content/drive', force_remount=True)

# Path presisi menuju folder spesifik kamu di Google Drive
TARGET_DIR = Path("/content/drive/MyDrive/YT_Online_Downloads")

# Folder internal khusus dependensi & environment
ENV_DIR = TARGET_DIR / "colab_env"
PKG_DIR = ENV_DIR / "python_packages"
BIN_DIR = ENV_DIR / "bin"
DOWNLOADS_DIR = TARGET_DIR / "downloads"
TEMP_DIR = TARGET_DIR / "temp"

# Buat seluruh struktur direktori secara otomatis jika belum ada
for folder in [TARGET_DIR, ENV_DIR, PKG_DIR, BIN_DIR, DOWNLOADS_DIR, TEMP_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

# ==============================================================================
# 2. INJEKSI SYSTEM PATH (MENGALIHKAN EXECUTION ENVIRONMENT KE DRIVE)
# ==============================================================================
# Memaksa Python membaca library/modul langsung dari Google Drive
if str(PKG_DIR) not in sys.path:
    sys.path.insert(0, str(PKG_DIR))

# Memaksa Linux terminal membaca aplikasi/binary langsung dari Google Drive
os.environ["PATH"] = f"{BIN_DIR}:{os.environ['PATH']}"
os.environ["PYTHONPATH"] = f"{PKG_DIR}:{os.environ.get('PYTHONPATH', '')}"

print(f"[+] Root Directory diset penuh ke: {TARGET_DIR}")

# ==============================================================================
# 3. INSTALASI DEPENDENCIES PERMANEN KE DALAM FOLDER DRIVE
# ==============================================================================
def setup_permanent_environment():
    print("\n[*] Memeriksa dan menginstal dependencies permanen ke Google Drive...")

    # Instal/Update modul Python langsung ke folder Drive
    subprocess.run([
        sys.executable, "-m", "pip", "install",
        "--target", str(PKG_DIR),
        "-U", "yt-dlp", "streamlink"
    ], stdout=subprocess.DEVNULL)

    # Mengunduh binary standalone yt-dlp ke folder bin di Drive
    ytdlp_bin = BIN_DIR / "yt-dlp"
    if not ytdlp_bin.exists():
        subprocess.run([
            "curl", "-L",
            "https://github.com/yt-dlp/yt-dlp/releases/latest/download/yt-dlp",
            "-o", str(ytdlp_bin)
        ], stdout=subprocess.DEVNULL)
        os.chmod(ytdlp_bin, 0o755)

    # Memastikan aplikasi pendukung sistem Linux terpasang
    apt_packages = ["ffmpeg", "aria2", "atomicparsley", "gpac", "mkvtoolnix", "nodejs"]
    subprocess.run(["apt-get", "update", "-qq"], stdout=subprocess.DEVNULL)
    subprocess.run(["apt-get", "install", "-y", "-qq"] + apt_packages, stdout=subprocess.DEVNULL)

    print("[+] Seluruh Environment & Dependencies tersimpan aman di Google Drive!\n")

setup_permanent_environment()
```
